```python
def marelig():
    gender=str(input("Your gender:"))
    age=int(input("Your age:"))
    if(age>=20 and gender=="male"):
        print("It is eligible")
        Marraige="It is eligible"
    elif(age>=18 and gender=="female"):
        print("It is eligible")
        Marraige="It is eligible"
    else:
        print("Not eligible")
        Marraige="Not eligible"
    return Marraige
marelig()
```
